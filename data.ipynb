## Important Question
Which Month Does Netflix Add the Most Content?

```python
df['month_added'] = df['date_added'].dt.month_name()

month_counts = df['month_added'].value_counts()

month_counts
```

```python
month_order = [
    'January','February','March','April','May','June',
    'July','August','September','October','November','December'
]

month_counts = df['month_added'].value_counts().reindex(month_order)

plt.figure(figsize=(10,5))
month_counts.plot(kind='bar')
plt.title('Content Added by Month')
plt.xlabel('Month')
plt.ylabel('Number of Titles')
plt.xticks(rotation=90)
plt.show()
```

## Observation
- July has the highest number of content additions on Netflix.
- December is the second-highest month for content additions.
- February has the lowest number of content additions.
- Content additions remain relatively consistent throughout most months of the year.
## Insight
- Netflix appears to increase content releases during the middle and end of the year.
- Higher content additions in July and December may be linked to holiday seasons and increased viewer activity.
- The platform follows a continuous content release strategy rather than concentrating releases in only a few months.
## Business Impact
- Understanding seasonal content addition patterns helps in planning content acquisition and marketing campaigns.
- Peak content release periods can be aligned with subscriber growth and engagement strategies.
